# Simple Summary Table
Want to show grade, loan amount, if it was paid or charged off. Limited to matured loans, as we want to extract all the historical data we can.

Brief insights:
- Grade A loans (less interest rate) appear more likely to be fully paid off compared to grade G loans.
- Higher interest loans seem to have much worse charge off rates; need to calculate whether or not the higher interest compensates.

```python
completed_loans = df[df['loan_status'].isin(['Fully Paid', 'Charged Off'])]

summary_table = completed_loans[['grade', 'loan_status']]
summary_table.groupby(['grade','loan_status']).size()
```
